## For loops

```python
# A Basic for loop
fruits = ['Apple', 'Banana', 'Cherry']

for fruit in fruits:
    print(fruit)
```

## Trying to replicate the max function using for loops

```python
# A data of student scores
scores = [78, 145, 92, 210, 65, 189, 233, 99, 156, 47,
          272, 118, 84, 199, 55, 300, 143, 91, 227, 68,
          175, 122, 89, 250, 73, 165, 566, 296, 61, 190,
          134, 97, 218, 59, 285]

# Assigning a base max score value
max_score = 0

# Using for loop to find the max score
for score in scores:
    if score > max_score:
        max_score = score
print(f"The maximum score is {max_score}")
```

## Using Range function in for loops

```python
# The range function does not include the last number
for num in range(1, 10):
    print(num)
```

```python
# Using step - step is nothing but how many numbers you would like to skip
for num in range(1, 10, 2):
    print(num)
```

```python
# A mini task - Fiding the sum of numbers from 1 to 100 using for loops and range
nums = 0

for num in range(1, 101):
    nums += num
print(nums)
```
