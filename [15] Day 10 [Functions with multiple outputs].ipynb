## Learning how to design functions with outputs 

```python
# .title() will convert the string to proper format
def format_name(f_name, l_name):
    formatted_f = f_name.title()
    formatted_l = l_name.title()

    return f'{formatted_f} {formatted_l}' # return is used to generate multiple outputs

print(format_name('BIll', 'ClInTon'))
```

```python
# Using multiple returns practically
# Desining a code that tells us if a year is leap year or not a leap year.

# Logic - 
# This is how you work out whether if a particular year is a leap year. 
# - on every year that is divisible by 4 with no remainder
# - except every year that is evenly divisible by 100 with no remainder 
# - unless the year is also divisible by 400 with no remainder   

def is_leap_year(year):
    if year % 4 == 0:
        if year % 100 != 0:
            return True
        elif year % 400 == 0:
            return True
        else:
            return False
    else:
        return False

is_leap_year(2026)
```
