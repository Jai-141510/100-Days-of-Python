```python
# Local scope - variable that we cannot use outside the function

def drink_portion():
    portion_strength = 2
    print(portion_strength)

portion_strength

# The below error states that it cannot identify a variable that is inside a function
```

```python
# Global scope - variable that we can use outside the function

player_health = 10

def drink_portion():
    portion_strength = 2
    print(player_health)

drink_portion()

# We get an output as the variable is outside the function
```
