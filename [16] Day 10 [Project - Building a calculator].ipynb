## Building a calculator using previously learnt functions and loops

```python
# Creating functions for each operation
def add (n1, n2):
    return n1 + n2

def subtract (n1, n2):
    return n1 - n2

def multiply (n1, n2):
    return n1 * n2

def divide (n1, n2):
    return n1 / n2


# Storing the operations in a dictionary
operations = {
    "+" : add,
    "-" : subtract,
    "*" : multiply,
    "/" : divide
}


should_accumulate = True
num1 = float(input("What's the first number?: "))

# Running a while loop to accumulate numbers if user clicks y
while should_accumulate:
    operation_symbol = input("Pick an operation +, -, /, *")
    num2 = float(input("What is the next number?: "))

    answer = operations[operation_symbol](num1, num2) # Using the dictionary and the function above to store the answer to a variable
    
    print(f"{num1} {operation_symbol} {num2} = {answer}") # Printing the answer for the user

    # Giving user the choice to continue with previous answer or reset
    choice = input("Press y to continue with the same number. Press n if you wish to restart the calculation: ")

    if choice == "y":
        num1 = answer
    else:
        should_accumulate = False







```
