## Dictionary

```python
# It contains a key value relationship within curly brackets
colors = {
    'apple': 'red', 
    'pear' : 'green', 
    'banana': 'yellow'
    }
```

```python
# This is how we will extract information from the dictionary
colors['pear']
```

```python
# This is how we will add values to a dictionary
colors['grapes'] = 'black'
colors
```

```python
# Wiping an existing dictionary
colors = {}
colors
```

```python
# Editing an existing dictionary
colors['apple'] = 'bright red'
colors
```

```python
# Looping through a dictionary - it will only loop through the keys and not the values
for val in colors:
    print(val)
```

```python
# To get the values after looping - we will pass the key value back to the dictionary
for val in colors:
    print(colors[val])
```

## Testing our knowledge - Creating a new dictionary for grades

```python
# Existing student scores
student_scores = {
    'Harry': 88,
    'Ron': 78,
    'Hermione': 95,
    'Draco': 75,
    'Neville': 60
}

# Creating an empty list which will be updated from the loop
student_grades = {}

# Creating a loop to assign grades
for student in student_scores:
    if student_scores[student] > 90:
        student_grades[student] = 'Outstanding'
    elif student_scores[student] > 80:
        student_grades[student] = 'Exceeds Expectations'
    elif student_scores[student] > 70:
        student_grades[student] = 'Acceptable'
    else:
        student_grades[student] = 'Fail'

print(student_grades)
```

# Designing a secret auction program

```python
userinput_name = input("What is your name?: ")
userinput_bid = input("Please enter your bid: ")

auction = {userinput_name : userinput_bid}

userinput_nextperson = input("Is there anyone after you? Type yes or no: ").lower()


while userinput_nextperson == 'yes':
    userinput_name2 = input("What is your name?: ")
    userinput_bid2 = input("Please enter your bid: ")
    auction[userinput_name2] = userinput_bid2
    userinput_nextperson = input("Is there anyone after you? Type yes or no: ").lower()

if userinput_nextperson == 'no':
    max_key = max(auction, key = auction.get)
    max_value = auction[max_key]
    print(f'{max_key} has the highest bid at ${max_value}')
```
