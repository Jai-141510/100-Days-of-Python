## Learning - Functions with inputs

```python
# Function with 1 input

def greet_with_name(name): # The value inside the paranthesis is called a Parameter
    print(f'Hello {name}')
    print(f'How do you do {name}?')


greet_with_name('User') # The input that you give when calling the function is called an Argument
```

```python
# Function with more than 1 input - Positional values and keyword values

def positional_greetings(name, location):
    print(f"Hello {name}")
    print(f"What is it like in {location}?")

positional_greetings('George', 'Alaska') # This is a positional argument, the way you enter values change the result


def keyword_greetings(name, location):
    print(f"Hello {name}")
    print(f"What is it like in {location}?")

keyword_greetings(location = 'Korea', name = 'Kim Min-ji')

```

## Building a Love calculator to reinforce the above concept

```python
# Rules - 
# 1. Take both people's names and check for the number of times the letters in the word TRUE occurs.
# 2. Then check for the number of times the letters in the word LOVE occurs.   
# 3. Then combine these numbers to make a 2 digit number and print it out. 

Word1 = 'true'
Word2 = 'love'



def calculate_love_score(name1, name2):
    Combinedname = name1.lower() + name2.lower()
    TrueCounter = 0
    LoveCounter = 0
    for n in Combinedname:
        if n in Word1:
            TrueCounter += 1
    for l in Combinedname:
        if l in Word2:
            LoveCounter += 1
    print(f'{TrueCounter}{LoveCounter}')
    


calculate_love_score(name1 = 'Kanye West', name2 = 'Kim Kardashian')
```
