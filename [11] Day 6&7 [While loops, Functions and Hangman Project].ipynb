## Hangman Project

```python
# Hangman stages
hangman_stages = [
    """
     ------
     |    |
     |    O
     |   /|\\
     |   / \\
     |
    ---
    """,
    """
     ------
     |    |
     |    O
     |   /|\\
     |   /
     |
    ---
    """,
    """
     ------
     |    |
     |    O
     |   /|\\
     |
     |
    ---
    """,
    """
     ------
     |    |
     |    O
     |   /|
     |
     |
    ---
    """,
    """
     ------
     |    |
     |    O
     |    |
     |
     |
    ---
    """,
    """
     ------
     |    |
     |    O
     |
     |
     |
    ---
    """,
    """
     ------
     |    |
     |
     |
     |
     |
    ---
    """
]
```

```python
# Creating a word bank
word_list = ["apple", "breeze", "candle", "dragon", "echo","forest", "glacier", "harbor", "island", "jungle", 
             "kettle", "lantern", "meadow", "nectar", "orbit","puzzle", "quartz", "river", "shadow", "thunder"]


# Letting the system choose a word
import random as rd
chosen_word = rd.choice(word_list)


# Printing out the underscores
placeholder = "" * len(chosen_word)
print(placeholder)

# Declaring variables to be updated using while loop
game_over = False # To determine when game will end
correct_letters = [] # To store letters for the display
lives = 7 # To store lives

# Inititating a while loop
while not game_over:
    guess = input("Guess a letter: ").lower() # Taking in user guess first

    # Dealing with the lives
    if guess not in chosen_word:
        lives -= 1
        print(hangman_stages[lives])
    else:
        correct_letters.append(guess) # updating the correct letters for display

    # Setting up the display code - this will show us the letters and underscores upon user input
    display = ""
    for letter in chosen_word:
        if letter == guess:
            display += letter
        elif letter in correct_letters:
            display += letter
        else:
            display += "_"
    
    print(display)

    # Setting up the exit code for the while loop - when lives are 0
    if lives == 0:
        game_over = True
        print("You lose")

    # Another exit code
    # for while loop - when user guesses all the words correctly
    if "_" not in display:
        game_over = True
        print("You win")

```
