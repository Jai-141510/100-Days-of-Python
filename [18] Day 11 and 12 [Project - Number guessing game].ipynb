## Creating a number guessing game

```python
# Asking for userinput for easy and hard
# easy - 10 tries and hard - 5 tries
# The range of the number will be 1-100
# each user guess will be compared to the guess number and will be marked low or high

import random as rd

print("Welcome to the Number Guessing Game!")
print("I'm thinking of a number between 1 and 100")

difficulty_level = input("Choose a difficulty. Type 'easy' or 'hard': ")

if difficulty_level == 'easy':
    num_tries = 10
    print("You have 10 tries, Good Luck!")
else:
    num_tries = 5
    print("You have 5 tries, Good Luck!")

system_num = rd.randint(1, 100)

game_over = False


while not game_over:

        if num_tries == 0:
            print("Game over! You have used all your attempts")
            print(f"The number was {system_num}")
            game_over = True
            break
        
        user_guess = int(input("Make a guess: "))

        if user_guess == system_num:
            print(f"That's correct! {user_guess} is the number!")
            game_over = True

        elif user_guess > system_num:
            num_tries -= 1
            print("Try guessing a lower number")
            print(f"You have {num_tries} tries")
        
        elif user_guess < system_num:
            num_tries -= 1
            print("Try guessing a higher number")
            print(f"You have {num_tries} tries ")
```
