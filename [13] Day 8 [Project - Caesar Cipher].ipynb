## Project - Building a Caesar Cipher

```python
# Creating a list of alphabets
alphabets = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

# Initiating a function
def Caesar_Cipher(option, message, shift):
    if option == 'encode':
        enc_message = "" # This will store the letter value as a string after each loop
        for n in message:
            position = alphabets.index(n) # This will get the index number of each letter in the word
            shifted_position = (position + shift) % 26 # We do this so that if number goes beyond Z, it will not crash
            enc_message += alphabets[shifted_position]
        print(enc_message)
    
    if option == 'decode':
        dec_message = "" # Storing decoded value as a string
        for d in message:
            position2 = alphabets.index(d) # This will get the index number of each letter in the word
            shifted_position2 = (position2 - shift) % 26 # We do this so that if number goes beyond Z, it will not crash
            dec_message += alphabets[shifted_position2]
        print(dec_message)


# Getting userinputs for the program
userinput_option = input("Type encode to encrypt or decode to decrypt: ") 
userinput_message = input("Type your message: ").lower()
userinput_shift = int(input("Type the shift number: "))


# Calling the function
Caesar_Cipher(option = userinput_option, message = userinput_message, shift = userinput_shift)
```
