```python
# Importing the random library
import random as rd
```

```python
# Randint is inclusive, it includes the first and the last number while generating random number
random_integer = rd.randint(1, 10)

# Testing out - generating a random number
random_integer
```

```python
# rd.random - it will include 0 but upto 1 - not including 1
random_num = rd.random() * 10

# Testing out - generating a random number
random_num
```

```python
# Trying to make a program that will print head or tail using random module

userinput_headortail = input("Do you want to go heads or tails?")

random_number = rd.randint(0, 1)

# Creating if statements
if random_number == 0:
    print(f"You predicted {userinput_headortail}, the coin is flipped as heads")
else:
    print(f"You predicted {userinput_headortail}, the coin is flipped as tails")
```
